```python
#Import Libraries
import nltk
import string

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
#Convert to Lowercase
text = """My name is Swaliha C.A.
I am from Aluva, Kerala. I completed my schooling at Government Higher Secondary School,Chowara.
There are four members in my family, including my parents, my sibling, and me.
I am currently pursuing my Bachelor of Technology in Computer Science and Artificial Intelligence at Adi Shankara Institute of Engineering and Technology.
My achievements include winning hackathons, being a Top 3 finalist in TinkerHer Hack 4.0, and completing internships in software development.
My hobbies are watching movies and listening to music.
I am interested in Artificial Intelligence, Machine Learning, Robotics, and Web Development.
My strengths are adaptable,listening teamwork, and problem-solving skills.
My weakness is that I sometimes spend too much time trying to make my work perfect and fear to communicate with people but i improved a lot than previous me."""
lower_text = text.lower()
print("Lowercase Text:\n")
print(lower_text)
```

```python
#Sentence Tokenization
import nltk
nltk.download('punkt_tab')
sentences = sent_tokenize(lower_text)

print("Sentence Tokenization:\n")
for sentence in sentences:
    print(sentence)
```

```python
#Word Tokenization
words = word_tokenize(lower_text)

print(words)
```

```python
#Remove Punctuation
words_no_punct = [word for word in words if word not in string.punctuation]

print(words_no_punct)
```

```python
#Remove Stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words_no_punct if word not in stop_words]

print(filtered_words)
```

```python
#Stemming
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print(stemmed_words)
```

```python
#Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print(lemmatized_words)
```

```python
#Final Preprocessed Text
final_text = " ".join(lemmatized_words)

print("Final Preprocessed Text:\n")
print(final_text)
```
